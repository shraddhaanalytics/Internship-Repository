```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:/Users/HP/Desktop/Internship Final Submission/NLP/2_raw_jobs.csv")
print(df.columns.tolist())
```

```python
# Basic Information
print(df.head())
print(df.info())
print(df.describe())
```

```python
total_number_of_jobs= len(df)
number_of_unique_job_titles=  df["job_title"].nunique()
most_common_job_roles =  df["job_title"].value_counts().head(10)

print("\nAnalyzing job dataset")
print("Total Number of jobs :", total_number_of_jobs)
print(" Number of unique job Titles:", number_of_unique_job_titles)
print("Top 10 most common job roles:",  most_common_job_roles)
```

```python
# Missing Value Analysis
df.isnull().sum()
print("Missing values present:", df.isnull().values.any())

print("\nMissing values in each column:")
print(df.isnull().sum())

if df.isnull().values.any():
    df = df.dropna()
    print("\nMissing values were removed.")
else:
    print("\nNo missing values found. No removal required.")

print("\nAfter cleaning:")
print("Missing values present:", df.isnull().values.any())
```

```python
# Duplicate data analysis
df.duplicated().sum()
print("Duplicate values present:",df.duplicated().values.any())
```

```python
# Character length
df["description_length"] = df["job_description"].str.len()

# Word length (optional, often more useful)
df["word_count"] = df["job_description"].str.split().str.len()

# Summary stats
print(df["description_length"].describe())
print(df["word_count"].describe())
```

```python
import matplotlib.pyplot as plt

most_common_job_roles.plot(kind="bar", figsize=(8, 2))

plt.title("Top 10 Most Common Job Title")
plt.xlabel("Job Title")
plt.ylabel("Number of Jobs")
#plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Number of jobs by location
jobs_by_location = df["location"].value_counts()

print("Jobs by location:")
print(jobs_by_location)

top_locations = df["location"].value_counts().head(10)

top_locations.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Job Locations")
plt.xlabel("Location")
plt.ylabel("Number of Jobs")
#plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Character length
df["description_length"] = df["job_description"].str.len()

# Word length (optional, often more useful)
df["word_count"] = df["job_description"].str.split().str.len()

# Summary stats
print(df["description_length"].describe())
print(df["word_count"].describe())

plt.hist(df["description_length"], bins=30)
plt.title("Description of Job  Distribution Length")
plt.xlabel("description_length")
plt.ylabel("Number of Jobs")
#plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
top_companies = df["company"].value_counts().head(10)

print(top_companies)

import matplotlib.pyplot as plt
#number_of_unique_companies=  df["company"].nunique()
plt.figure(figsize=(6, 3))

top_companies.sort_values().plot(kind="barh")

plt.title("Top 10 Companies by Number of Jobs")
plt.xlabel("Number of Jobs")
plt.ylabel("Company")

plt.tight_layout()
plt.show()
```
