```python
import numpy as np
import pandas as pd
import random
import re

# Read dataset
df = pd.read_csv("C:/Users/HP/Desktop/Internship Final Submission/NLP/2_raw_jobs.csv")
```

```python
# 1. Define Alias & Canonical Dictionary
# Maps variations (lowercase & cleaned) to the canonical skill name
ALIAS_MAP = {
    # Python variations
    "python": "Python",
    "python3": "Python",
    "python 3": "Python",
    "python programming": "Python",
    
    # Power BI variations
    "powerbi": "Power BI",
    "power bi": "Power BI",
    "power-bi": "Power BI",
    
    # PostgreSQL variations
    "postgres": "PostgreSQL",
    "postgresql": "PostgreSQL",
    "ms sql": "SQL Server",
    "mssql": "SQL Server",
    
    # Machine Learning / AI variations
    "ml": "Machine Learning",
    "machine learning": "Machine Learning",
    "scikit learn": "scikit-learn",
    "scikit-learn": "scikit-learn",
    
    # Cloud & DevOps
    "aws": "AWS",
    "aws cloud": "AWS",
    "k8s": "Kubernetes",
    "kubernetes": "Kubernetes"
}
```

```python
# 2. Pipeline Processing Steps[cite: 4]

def clean_text(text: str) -> str:
    """Step 1 & 2: Lowercase & remove unwanted special characters[cite: 4]"""
    if pd.isna(text):
        return ""
    text = str(text).lower().strip()
    # Remove special characters except spaces and hyphens[cite: 4]
    text = re.sub(r"[^\w\s-]", "", text)
    # Collapse multiple spaces
    text = re.sub(r"\s+", " ", text)
    return text

def normalize_skill(raw_skill: str) -> str:
    """Step 3 & 4: Map cleaned raw skills to Canonical Skills[cite: 4]"""
    cleaned = clean_text(raw_skill)
    
    # Direct lookup in Alias Mapping[cite: 4]
    if cleaned in ALIAS_MAP:
        return ALIAS_MAP[cleaned]
    
    # Fallback/Default: Capitalize words if no canonical match exists
    return raw_skill.strip().title()
```

```python
# ==========================================
# 3. Batch Normalization Engine Execution
# ==========================================
def run_normalization_pipeline(input_skills: list) -> pd.DataFrame:
    records = []
    
    for raw in input_skills:
        cleaned = clean_text(raw)
        canonical = normalize_skill(raw)
        records.append({
            "raw_skill": raw,
            "cleaned_skill": cleaned,
            "canonical_skill": canonical
        })
        
    df = pd.DataFrame(records)
    return df

```

```python
# 4. Main Execution & CSV Export
if __name__ == "__main__":
    # Test sample containing variations from the Day 17 specification[cite: 4]
    sample_raw_skills = [
        "python3",
        "Python 3",
        "Python programming",
        "PowerBI",
        "power-bi",
        "Power BI",
        "postgres",
        "PostgreSQL",
        "AWS Cloud",
        "k8s",
        "scikit learn"
    ]
    
    # Process skills through pipeline
    result_df = run_normalization_pipeline(sample_raw_skills)
    
    # Display output
    print("=== Skill Normalization Sample Result ===")
    print(result_df[["raw_skill", "canonical_skill"]])
    
    # Export final deliverable: normalized_skills.csv[cite: 4]
    output_filename = "normalized_skills.csv"
    result_df.to_csv(output_filename, index=False)
    print(f"\nSuccessfully exported deliverable to '{output_filename}'")
```
